## Our task: machine translation

We gonna try our encoder-decoder models on russian to english machine translation problem. More specifically, we'll translate hotel and hostel descriptions. This task shows the scale of machine translation while not requiring you to train your model for weeks if you don't use GPU.

Before we get to the architecture, there's some preprocessing to be done. ~~Go tokenize~~ Alright, this time we've done preprocessing for you. As usual, the data will be tokenized with WordPunctTokenizer.

However, there's one more thing to do. Our data lines contain unique rare words. If we operate on a word level, we will have to deal with large vocabulary size. If instead we use character-level models, it would take lots of iterations to process a sequence. This time we're gonna pick something inbetween.

One popular approach is called [Byte Pair Encoding](https://github.com/rsennrich/subword-nmt) aka __BPE__. The algorithm starts with a character-level tokenization and then iteratively merges most frequent pairs for N iterations. This results in frequent words being merged into a single token and rare words split into syllables or even characters.

### Encoder-decoder model

The code below contains a template for a simple encoder-decoder model: single GRU encoder/decoder, no attention or anything. This model is implemented for you as a reference and a baseline for your homework assignment.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(device)
```

```python
class BasicModel(nn.Module):
    def __init__(self, inp_voc, out_voc, emb_size=64, hid_size=128):
        """
        A simple encoder-decoder seq2seq model
        """
        super().__init__() # initialize base class to track sub-layers, parameters, etc.

        self.inp_voc, self.out_voc = inp_voc, out_voc
        self.hid_size = hid_size

        self.emb_inp = nn.Embedding(len(inp_voc), emb_size) # Это обучаемые вектора эмбеддингов, которые будут менять в процессе обучения
        self.emb_out = nn.Embedding(len(out_voc), emb_size)
        self.enc0 = nn.GRU(emb_size, hid_size, batch_first=True) # Обрабатывает сразу последовательность токенов

        self.dec_start = nn.Linear(hid_size, hid_size)
        self.dec0 = nn.GRUCell(emb_size, hid_size) # Обрабатывает по одному токену
        self.logits = nn.Linear(hid_size, len(out_voc))

    def forward(self, inp, out):
        """ Apply model in training mode """
        initial_state = self.encode(inp)
        return self.decode(initial_state, out)


    def encode(self, inp, **flags):
        """
        Takes symbolic input sequence, computes initial state
        :param inp: matrix of input tokens [batch, time]
        :returns: initial decoder state tensors, one or many
        """
        inp_emb = self.emb_inp(inp)
        batch_size = inp.shape[0]

        enc_seq, [last_state_but_not_really] = self.enc0(inp_emb)
        # enc_seq: [batch, time, hid_size], last_state: [batch, hid_size]

        # note: last_state is not _actually_ last because of padding, let's find the real last_state
        lengths = (inp != self.inp_voc.eos_ix).to(torch.int64).sum(dim=1).clamp_max(inp.shape[1] - 1)
        last_state = enc_seq[torch.arange(len(enc_seq)), lengths]
        # ^-- shape: [batch_size, hid_size]

        dec_start = self.dec_start(last_state)
        return [dec_start]

    def decode_step(self, prev_state, prev_tokens, **flags):
        """
        Takes previous decoder state and tokens, returns new state and logits for next tokens
        :param prev_state: a list of previous decoder state tensors, same as returned by encode(...)
        :param prev_tokens: previous output tokens, an int vector of [batch_size]
        :return: a list of next decoder state tensors, a tensor of logits [batch, len(out_voc)]
        """
        prev_gru0_state = prev_state[0]
        out_emb = self.emb_out(prev_tokens)
        new_dec_state = self.dec0(out_emb, prev_gru0_state)
        output_logits = self.logits(new_dec_state)

        return [new_dec_state], output_logits

    def decode(self, initial_state, out_tokens, **flags):
        """ Iterate over reference tokens (out_tokens) with decode_step
        :param initial_state: a list of initial decoder state tensors, same as returned by encode(...)
        :param out_tokens: an int matrix of [batch, time]
        :return: a matrix of logits for each time
        """
        batch_size = out_tokens.shape[0]
        state = initial_state

        # initial logits: always predict BOS
        onehot_bos = F.one_hot(torch.full([batch_size], self.out_voc.bos_ix, dtype=torch.int64),
                               num_classes=len(self.out_voc)).to(device=out_tokens.device)
        first_logits = torch.log(onehot_bos.to(torch.float32) + 1e-9)

        logits_sequence = [first_logits]
        for i in range(out_tokens.shape[1] - 1):
            state, logits = self.decode_step(state, out_tokens[:, i])
            logits_sequence.append(logits)
        return torch.stack(logits_sequence, dim=1)

    def decode_inference(self, initial_state, max_len=100, **flags):
        """ Generate translations from model (greedy version)
        :return: a list of vocabulary ids to token [max_len, batch_size]
        """
        batch_size, device = len(initial_state[0]), initial_state[0].device
        state = initial_state
        outputs = [torch.full([batch_size], self.out_voc.bos_ix, dtype=torch.int64,
                              device=device)]
        all_states = [initial_state]

        for i in range(max_len):
            state, logits = self.decode_step(state, outputs[-1])
            outputs.append(logits.argmax(dim=-1))
            all_states.append(state)

        return torch.stack(outputs, dim=1), all_states

    def translate_lines(self, inp_lines, **kwargs):
        inp = self.inp_voc.to_matrix(inp_lines).to(device)
        initial_state = self.encode(inp)
        out_ids, states = self.decode_inference(initial_state, **kwargs)
        return self.out_voc.to_lines(out_ids.cpu().numpy()), states
```

### Training loss (2 points)

Our training objective is almost the same as it was for neural language models:
$$ L = {\frac1{|D|}} \sum_{X, Y \in D} \sum_{y_t \in Y} - \log p(y_t \mid y_1, \dots, y_{t-1}, X, \theta) $$

where $|D|$ is the __total length of all sequences__, including BOS and first EOS, but excluding PAD.

```python
def compute_loss(model, inp, out, **flags):
    """
    Compute loss (float32 scalar) as in the formula above
    :param inp: input tokens matrix, int32[batch, time]
    :param out: reference tokens matrix, int32[batch, time]

    In order to pass the tests, your function should
    * include loss at first EOS but not the subsequent ones
    * divide sum of losses by a sum of input lengths (use voc.compute_mask)
    """
    mask = model.out_voc.compute_mask(out) # [batch_size, out_len]
    targets_1hot = F.one_hot(out, len(model.out_voc)).to(torch.float32)

    # outputs of the model, [batch_size, out_len, num_tokens]
    logits_seq = model.forward(inp, out) * mask.unsqueeze(-1)
    # log-probabilities of all tokens at all steps, [batch_size, out_len, num_tokens]
    logprobs_seq = torch.log(F.softmax(logits_seq, dim=-1))
    # log-probabilities of correct outputs, [batch_size, out_len]
    logp_out = (logprobs_seq * targets_1hot).sum(dim=-1)
    logp_out = torch.where(mask, logp_out, 0) # Не считаем паддинг
    # ^-- this will select the probability of the actual next token.
    # Note: you can compute loss more efficiently using using F.cross_entropy

    # average cross-entropy over tokens where mask == True
    return -logp_out.sum() / mask.sum()  # average loss, scalar
```

### Evaluation: BLEU

Machine translation is commonly evaluated with [BLEU](https://en.wikipedia.org/wiki/BLEU) score. This metric simply computes which fraction of predicted n-grams is actually present in the reference translation. It does so for n=1,2,3 and 4 and computes the geometric average with penalty if translation is shorter than reference.

While BLEU [has many drawbacks](http://www.cs.jhu.edu/~ccb/publications/re-evaluating-the-role-of-bleu-in-mt-research.pdf), it still remains the most commonly used metric and one of the simplest to compute.

```python
from nltk.translate.bleu_score import corpus_bleu
def compute_bleu(model, inp_lines, out_lines, bpe_sep='@@ ', **flags):
    """
    Estimates corpora-level BLEU score of model's translations given inp and reference out
    Note: if you're serious about reporting your results, use https://pypi.org/project/sacrebleu
    """
    with torch.no_grad():
        translations, _ = model.translate_lines(inp_lines, **flags)
        translations = [line.replace(bpe_sep, '') for line in translations]
        actual = [line.replace(bpe_sep, '') for line in out_lines]
        return corpus_bleu(
            [[ref.split()] for ref in actual],
            [trans.split() for trans in translations],
            smoothing_function=lambda precisions, **kw: [p + 1.0 / p.denominator for p in precisions]
            ) * 100
```

### Training loop

Training encoder-decoder models isn't that different from any other models: sample batches, compute loss, backprop and update

```python
from IPython.display import clear_output
from tqdm import tqdm, trange
metrics = {'train_loss': [], 'dev_bleu': [] }

model = BasicModel(inp_voc, out_voc).to(device)
opt = torch.optim.Adam(model.parameters(), lr=1e-3)
batch_size = 32
```

```python
for _ in trange(25000):
    step = len(metrics['train_loss']) + 1
    batch_ix = np.random.randint(len(train_inp), size=batch_size)
    batch_inp = inp_voc.to_matrix(train_inp[batch_ix]).to(device)
    batch_out = out_voc.to_matrix(train_out[batch_ix]).to(device)

    loss_t = compute_loss(model, batch_inp, batch_out)
    loss_t.backward()
    opt.step()
    opt.zero_grad()

    metrics['train_loss'].append((step, loss_t.item()))

    if step % 100 == 0:
        metrics['dev_bleu'].append((step, compute_bleu(model, dev_inp, dev_out)))

        clear_output(True)
        plt.figure(figsize=(12,4))
        for i, (name, history) in enumerate(sorted(metrics.items())):
            plt.subplot(1, len(metrics), i + 1)
            plt.title(name)
            plt.plot(*zip(*history))
            plt.grid()
        plt.show()
        print("Mean loss=%.3f" % np.mean(metrics['train_loss'][-10:], axis=0)[1], flush=True)

# Note: it's okay if bleu oscillates up and down as long as it gets better on average over long term (e.g. 5k batches)
```

### Scaled Dot Product (3 Points)
Here you will have to implement a layer that computes a **scaled dot-product attention**:

Given encoder sequence $h^e_0, h^e_1, h^e_2, \dots, h^e_T$ and a single decoder state $h^d$,

* Project the decoder state into a **query** vector
  $$q = W_q h^d$$
* Project encoder states into **keys** and **values**
  $$k_t = W_k h^e_t, \quad v_t = W_v h^e_t$$
* Compute logits as **scaled dot products** between the query and each key
  $$a_t = \frac{q \cdot k_t}{\sqrt{d_k}}$$
  where $d\_k$ is the key/query dimensionality.
* Convert logits into probabilities using softmax
  $$p_t = \frac{e^{a_t}}{\sum_\tau e^{a_\tau}}$$
* Add up encoder values with probabilities to get the **attention response**
  $$attn = \sum_t p_t \cdot v_t$$

```python
from math import sqrt

class DotProductAttentionLayer(nn.Module):
    def __init__(self, name,
                 enc_size, dec_size,
                 emb_size,
                #  hid_size,
                 attn_size):
        """
        Scaled dot-product attention with projections
        :param enc_size: num units in encoder state
        :param dec_size: num units in decoder state
        :param hid_size: dimension of query/key space (d_k)
        """
        super().__init__()
        self.name = name
        self.enc_size = enc_size
        self.dec_size = dec_size
        # self.hid_size = hid_size
        self.attn_size = attn_size

        # Define projection layers
        # W_q, W_k, W_v project decoder states into query, key, value
        self.W_q = nn.Linear(dec_size, attn_size) # [batch_size x hid_size]
        self.W_k = nn.Linear(emb_size, attn_size) # [batch_size x ninp x hid_size]
        self.W_v = nn.Linear(emb_size, attn_size) # [batch_size x ninp x hid_size]

    def forward(self, enc, dec, inp_mask):
        """
        Compute attention response and weights
        :param enc: encoder sequence [batch_size, ninp, enc_size]
        :param dec: decoder state [batch_size, dec_size]
        :param inp_mask: mask [batch_size, ninp] (0 = padding)
        :returns: attn [batch_size, enc_size], probs [batch_size, ninp]
        """
        # print(f"---- attn ----")

        # Compute Q, K, V projections

        # print("dec", dec.shape)
        # print("enc", enc.shape)
        Q = self.W_q(dec)
        K = self.W_k(enc)
        V = self.W_v(enc)

        # Compute logits as scaled dot product QK^T, do not forget 1 / d
        # Hint: You may want to expand Q to shape [batch_size, 1, hid_size]
        logits = torch.matmul(Q.unsqueeze(dim=1), torch.transpose(K, 2, 1)) / sqrt(self.attn_size)
        logits = logits.squeeze(dim=1)
        # print("logits", logits.shape) # [batch_size x ninp]

        # Apply mask - if mask is 0, logits should be -inf or -1e9
        # You may need torch.where
        logits = torch.where(inp_mask, logits, 0) # [batch_size x ninp]

        # Softmax
        probs = F.softmax(logits, dim=-1) # [batch_size x ninp]

        # Compute attention response
        # print("probs", probs.shape)
        # print("V", V.shape)
        attn = (probs.unsqueeze(dim=-1) * V).sum(dim=1) # [batch_size x hid_size]
        # print("attn", attn.shape)
        # print(attn.shape, probs.shape)
        return attn, probs
```

### Seq2seq model with attention (3 points)

You can now use the attention layer to build a network. The simplest way to implement attention is to use it in decoder phase:
![img](https://i.imgur.com/6fKHlHb.png)
_image from distill.pub [article](https://distill.pub/2016/augmented-rnns/)_

On every step, use __previous__ decoder state to obtain attention response. Then feed concat this response to the inputs of next attention layer.

The key implementation detail here is __model state__. Put simply, you can add any tensor into the list of `encode` outputs. You will then have access to them at each `decode` step. This may include:
* Last RNN hidden states (as in basic model)
* The whole sequence of encoder outputs (to attend to) and mask
* Attention probabilities (to visualize)

_There are, of course, alternative ways to wire attention into your network and different kinds of attention. Take a look at [this](https://arxiv.org/abs/1609.08144), [this](https://arxiv.org/abs/1706.03762) and [this](https://arxiv.org/abs/1808.03867) for ideas. And for image captioning/im2latex there's [visual attention](https://arxiv.org/abs/1502.03044)_

```python
class AttentiveModel(BasicModel): # Luong style model
    def __init__(self, name, inp_voc, out_voc,
                 emb_size=64,
                 enc_size=128,
                 dec_size=256,
                #  hid_size=128,
                 attn_size=128, enc_layers=1):
        """ Translation model that uses attention. See instructions above. """
        nn.Module.__init__(self)  # initialize base class to track sub-layers, trainable variables, etc.
        self.inp_voc, self.out_voc = inp_voc, out_voc
        # self.hid_size = hid_size
        # self.emb_size = emb_size
        # self.hid_size = hid_size
        # self.attn_size = attn_size
        # self.name = name

        # trainable embeddings
        self.inp_emb = nn.Embedding(len(inp_voc), emb_size)
        self.out_emb = nn.Embedding(len(out_voc), emb_size)

        # n-layer RNN encoder
        self.enc0 = nn.GRU(emb_size, enc_size, num_layers=enc_layers, batch_first=True)

        # attention
        self.attn = DotProductAttentionLayer(name, enc_size=enc_size, dec_size=dec_size, emb_size=emb_size, attn_size=attn_size)

        # n layers of RNN cell in decoder
        self.dec0 = nn.GRUCell(emb_size, dec_size)
        # self.dec1 = nn.GRUCell(emb_size, hid_size)
        # self.dec2 = nn.GRUCell(emb_size, hid_size)

        # trasform concated attetion and decoder hidden state to one vector of hid_size
        self.create_new_prev_state = nn.Linear(attn_size + dec_size, dec_size)

        # final logits
        self.logits = nn.Linear(dec_size, len(out_voc))
        self.transform_enc_out_to_dec_inp = nn.Linear(enc_size, dec_size)


    def encode(self, inp, **flags):
        """
        Takes symbolic input sequence, computes initial state
        :param inp: matrix of input tokens [batch, time]
        :return: a list of initial decoder state tensors
        """
        # print(f"---- encode ----")
        # encode input sequence, create initial decoder states
        inp_mask = (inp != self.inp_voc.eos_ix) # [batch_size x ninp]
        lengths = (inp != self.inp_voc.eos_ix).to(torch.int64).sum(dim=1) # [batch_size]
        input_sequence = self.inp_emb(inp)
        enc_states, last_state_but_not_real = self.enc0(input_sequence) # [batch_size x time x enc_size]

        # We want take the real last state, not padding last state
        last_states = enc_states[torch.arange(len(enc_states)), lengths] # [batch_size x enc_size]
        dec_start = self.transform_enc_out_to_dec_inp(last_states)

        # apply attention layer from initial decoder hidden state
        # print("input_sequence", input_sequence.shape)
        # print("last_states", dec_start.shape)
        # print("inp_mask", inp_mask.shape)

        # first_attn_probas = self.attn(input_sequence, dec_start, inp_mask)
        # print("first_attn_probas", first_attn_probas.shape)

        # Build first state: include
        # * initial states for decoder recurrent layers
        # * encoder sequence and encoder attn mask (for attention)
        # * make sure that last state item is attention probabilities tensor
        # print(dec_start.shape, first_attn_probas[0].shape, first_attn_probas[1].shape)
        # first_state = [dec_start]
        return [dec_start, input_sequence, inp_mask]

    def decode_step(self, prev_state, prev_tokens, **flags):
        """
        Takes previous decoder state and tokens, returns new state and logits for next tokens
        :param prev_state: a list of previous decoder state tensors
        :param prev_tokens: previous output tokens, an int vector of [batch_size]
        :return: a list of next decoder state tensors, a tensor of logits [batch, n_tokens]
        """
        # print(f"---- decode step ----")
        # I think prev_state is a combination of attention and hidden state from previos layer

        new_dec_state = self.dec0(self.out_emb(prev_tokens), prev_state) # [batch_size x dec_size]
        output_logits = self.logits(new_dec_state)
        # print(new_dec_state.shape, output_logits.shape)
        return [new_dec_state, output_logits]

    def decode(self, enc_init_state, output_tokens, **flags):
        """
        Takes encoder states, output tokens process its through the decoder with attention.
        :param enc_states: a list of decoder initial states and attention inital state and probs
        :param output_tokens: target tokens that will be masked. [batch_size x nout]
        :return: output_logits [batch_size x len(out_voc)]
        """
        # print(f"---- decode ----")
        batch_size = output_tokens.shape[0]
        dec_start, input_sequence, inp_mask = enc_init_state
        # concat attention and previous decoder hidden state to create prev_state
        # pass it through the fc layer
        # print("enc_state[0]", enc_state[0].shape)
        # print("enc_state[1]", enc_state[1].shape)

        first_attn,  _ = self.attn(input_sequence, dec_start, inp_mask)
        new_prev_state = self.create_new_prev_state(torch.concat((dec_start, first_attn), 1)) # [batch_size x dec_size]

        # we want BOS as a first token of the model prediction


        first_output = F.one_hot(torch.full([batch_size], self.out_voc.bos_ix, dtype=torch.int64),
                               num_classes=len(self.out_voc)).to(device=output_tokens.device)
        first_logits = torch.log(first_output.to(torch.float32) + 1e-9)
        output_logits = [first_logits]

        # encode input sequence, create initial decoder states
        # inp_mask = (input_tokens != self.inp_voc.eos_ix) # [batch_size x ninp]
        # lengths = (input_tokens != self.inp_voc.eos_ix).to(torch.int64).sum(dim=1) # [batch_size]
        # input_sequence = self.inp_emb(input_tokens)

        # iterate through the output tokens
        for i in range(output_tokens.shape[1] - 1):
            prev_dec_state, logits = self.decode_step(new_prev_state, output_tokens[:, i])
            attention, _ = self.attn(input_sequence, prev_dec_state, inp_mask) # add attention
            new_prev_state = self.create_new_prev_state(torch.concat((prev_dec_state, attention), 1))
            output_logits.append(logits)

        return torch.stack(output_logits, dim=1)

    def decode_inference(self, enc_init_state, input_tokens, max_len=100, **flags):
        """ Generate translations from model (greedy version)
        :return: a list of vocabulary ids to token [max_len, batch_size]
        """

        batch_size = input_tokens.shape[0]
        device = input_tokens.device
        dec_start, input_sequence, inp_mask = enc_init_state

        # concat attention and previous decoder hidden state to create prev_state
        # pass it through the fc layer
        first_attn, first_probs = self.attn(input_sequence, dec_start, inp_mask)
        new_prev_state = self.create_new_prev_state(torch.concat((dec_start, first_attn), 1)) # [batch_size x dec_size]

        # we want BOS as a first token of the model prediction
        output_tokens = [torch.full([batch_size], self.out_voc.bos_ix, dtype=torch.int64).to(device)]
        all_states = [first_probs]

        # encode input sequence, create initial decoder states
        # inp_mask = (input_tokens != self.inp_voc.eos_ix) # [batch_size x ninp]
        # lengths = (input_tokens != self.inp_voc.eos_ix).to(torch.int64).sum(dim=1) # [batch_size]
        # input_sequence = self.inp_emb(input_tokens)

        # iterate through the output tokens
        for i in range(max_len - 1):
            prev_dec_state, logits = self.decode_step(new_prev_state, output_tokens[-1])
            attention, probs = self.attn(input_sequence, prev_dec_state, inp_mask) # add attention
            new_prev_state = self.create_new_prev_state(torch.concat((prev_dec_state, attention), 1)) # [batch_size x dec_size]

            # EDA
            # print(logits[:, :5])

            # print(new_prev_state.shape)
            output_tokens.append(logits.argmax(dim=-1))
            all_states.append(probs)
        return torch.stack(output_tokens, dim=1), all_states


    def translate_lines(self, inp_lines, **kwargs):
        inp = self.inp_voc.to_matrix(inp_lines).to(device)
        # print(inp.shape)
        enc_init_state = self.encode(inp)
        out_ids, states = self.decode_inference(enc_init_state, inp, **kwargs)
        return self.out_voc.to_lines(out_ids.cpu().numpy()), states

    def beam_translate(self, text, beam_size=3, max_len=100, **kwargs):
        """Generate beam_size translation variants
        :param text: a string of input text (need tokenization)
        :return: a beam_size legth translation list
        """
        text = ' '.join(self.inp_voc.tokenize(text.lower()))
        inp_lines = [text] * beam_size
        inp = self.inp_voc.to_matrix(inp_lines).to(device)
        enc_init_state = self.encode(inp)

        # make first attention
        dec_start, input_sequence, inp_mask = self.encode(inp)
        first_attn, _ = self.attn(input_sequence, dec_start, inp_mask)
        new_prev_state = self.create_new_prev_state(torch.concat((dec_start, first_attn), 1)) # [beam_size x dec_size]

        # start BOS prediciton
        output_tokens = [torch.full([beam_size], self.out_voc.bos_ix, dtype=torch.int64).to(device)]
        beams = [("_BOS_", 1) * beam_size] # need to compare probs

        # Predict first different tokens after BOS
        prev_dec_state, logits = self.decode_step(new_prev_state, output_tokens[-1])
        attention, _ = self.attn(input_sequence, prev_dec_state, inp_mask) # add attention
        new_prev_state = self.create_new_prev_state(torch.concat((prev_dec_state, attention), 1)) # [batch_size x dec_size]
        probs = torch.softmax(logits[0].unsqueeze(0), dim=-1).detach().cpu().numpy()
        top_tokens_flattened = np.argpartition(probs, -beam_size, axis=None)[-beam_size:]
        beam_ids = top_tokens_flattened // len(self.out_voc)
        top_tokens = top_tokens_flattened % len(self.out_voc)
        sorted_tokens = sorted(tuple(zip(beam_ids, top_tokens)), key=lambda x: x[0])
        output_tokens.append(torch.tensor([token_id for (_, token_id) in sorted_tokens]).to(device))

        # Continue translate
        for i in range(max_len - 2):
            prev_dec_state, logits = self.decode_step(new_prev_state, output_tokens[-1])
            attention, _ = self.attn(input_sequence, prev_dec_state, inp_mask) # add attention
            new_prev_state = self.create_new_prev_state(torch.concat((prev_dec_state, attention), 1)) # [batch_size x dec_size]

            # keep beam_size variants
            probs = torch.softmax(logits, dim=-1).detach().cpu().numpy()
            top_tokens_flattened = np.argpartition(probs, -beam_size, axis=None)[-beam_size:]
            beam_ids = top_tokens_flattened // len(self.out_voc)
            top_tokens = top_tokens_flattened % len(self.out_voc)
            sorted_tokens = sorted(tuple(zip(beam_ids, top_tokens)), key=lambda x: x[0])
            output_tokens.append(torch.tensor([token_id for (_, token_id) in sorted_tokens]).to(device))

        return self.out_voc.to_lines(torch.stack(output_tokens, dim=1).cpu().numpy())

    def forward(self, inp, out):
        enc_init_state = self.encode(inp)
        return self.decode(enc_init_state, out)
```

```python
from nltk.translate.bleu_score import corpus_bleu
def compute_bleu(model, inp_lines, out_lines, bpe_sep='@@ ', batch_size=64, **flags):
    """
    Add batching
    Estimates corpora-level BLEU score of model's translations given inp and reference out
    Note: if you're serious about reporting your results, use https://pypi.org/project/sacrebleu
    """
    with torch.no_grad():
        all_translations = []
        for i in range(0, len(inp_lines), batch_size):
            batch_inp = inp_lines[i:i + batch_size]
            translations_batch, _ = model.translate_lines(batch_inp, **flags)
            all_translations.extend(translations_batch)

        translations = [line.replace(bpe_sep, '') for line in all_translations]
        actual = [line.replace(bpe_sep, '') for line in out_lines]
        return corpus_bleu(
            [[ref.split()] for ref in actual],
            [trans.split() for trans in translations],
            smoothing_function=lambda precisions, **kw: [p + 1.0 / p.denominator for p in precisions]
            ) * 100
```

Here you will have to implement a layer that computes a simple additive attention:

Given encoder sequence $ h^e_0, h^e_1, h^e_2, ..., h^e_T$ and a single decoder state $h^d$,

* Compute logits with a 2-layer neural network
$$a_t = linear_{out}(tanh(linear_{e}(h^e_t) + linear_{d}(h_d)))$$
* Get probabilities from logits,
$$ p_t = {{e ^ {a_t}} \over { \sum_\tau e^{a_\tau} }} $$

* Add up encoder states with probabilities to get __attention response__
$$ attn = \sum_t p_t \cdot h^e_t $$
